```python
import pandas as pd
e_path = "../dataset/raw/en_energy_extraversion.json"
i_path = "../dataset/raw/en_energy_introversion.json"
```

```python
data_e = pd.read_json(e_path)
data_i = pd.read_json(i_path)
print(data_e.head())
```

```python
data_e = data_e[['output']]
data_e.rename(columns={'output': 'text'}, inplace=True)

data_i = data_i[['output']]
data_i.rename(columns={'output': 'text'}, inplace=True)

data_e['label'] = 1  # 为“e人”数据添加标签1
data_i['label'] = 0  # 为“i人”数据添加标签0

data = pd.concat([data_e, data_i], ignore_index=True)

data = data.sample(frac=1).reset_index(drop=True)

data.to_csv('../dataset/processed/combined_ie_dataset.csv', index=False, encoding='utf-8')
```
